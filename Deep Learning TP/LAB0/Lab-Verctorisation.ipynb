## 2. Opérations Élément par Élément (Vectorisation I)

Ces opérations s'appliquent à tous les éléments d'un tableau sans boucle, ce qui est essentiel pour les fonctions d'activation et le calcul du coût.

### Exercice 2.1 : Calcul Vectorisé d'une Expression

Soit le vecteur d'entrée $X = [1, 2, 3, 4]$. Calculez le vecteur de sortie $Y$ tel que :
$$
Y = \frac{1}{2}X^3 + 10
$$ 
Utilisez uniquement des opérations NumPy (pas de boucles `for`).

```python
X = np.array([1, 2, 3, 4])

# Votre calcul de Y ici
Y = 0.5 * (X ** 3) + 10

print(f"X:\t{X}")
print(f"Y:\t{Y}")
```

## 3. Le Cœur du Deep Learning : Produit Matriciel (Vectorisation II)

Le calcul de l'entrée pondérée d'un neurone est $Z = WX + b$. Nous pratiquons ici l'opération centrale : $W \cdot X$.

### Exercice 3.1 : Produit Matrice-Vecteur

Soit la matrice de poids $W$ et le vecteur d'entrée $X$ :

| Taille | Description |
| :--- | :--- |
| $W$: 2x3 | 2 neurones, 3 entrées |
| $X$: 3x1 | 3 entrées |

Calculez le résultat $Z = W \cdot X$ en utilisant l'opérateur **`@`** ou la fonction `np.dot`.

```python
# Matrice de Poids W (2x3)
W = np.array([[0.5, 0.1, -0.2],
              [0.3, 0.9, 0.4]])

# Vecteur d'Entrée X (3x1)
X = np.array([10, 5, 2])

# Calculez Z = W @ X
Z = W @ X

print(f"Matrice W (2x3):\n{W}")
print(f"\nVecteur X (3x1):\n{X}")
print(f"\nRésultat Z = W @ X (2x1):\n{Z}")
```

## 4. Application DL : Vectorisation de la fonction Sigmoïde

La fonction d'activation **Sigmoïde** est $\sigma(z) = \frac{1}{1 + e^{-z}}$. Elle doit être appliquée à un ensemble de valeurs $Z$ pour être utilisable.

### Exercice 4.1 : Implémentation Vectorisée

Définissez une fonction `sigmoid_vectorisee(Z)` qui prend en entrée un tableau NumPy `Z` et retourne le résultat de $\sigma(Z)$ en utilisant `np.exp` pour la vectorisation. Testez-la sur la matrice `Z_matrice` fournie.

```python
def sigmoid_vectorisee(Z):
    """Calcule la fonction Sigmoïde de manière vectorisée."""
    # Votre code ici
    return 1 / (1 + np.exp(-Z))

# Matrice de test (valeurs d'entrée brutes d'un neurone)
Z_matrice = np.array([[-1.0, 0.0, 1.0],
                      [5.0, -5.0, 0.5]])

# Calculez A = sigmoid(Z)
A = sigmoid_vectorisee(Z_matrice)

print(f"Matrice Z (Entrée):\n{Z_matrice}")
print(f"\nMatrice A (Sortie Sigmoïde):\n{A}")
```

## 5. Mesure de la Performance : L'intérêt de la Vectorisation

La vectorisation est la clé de la **performance** en DL. Nous allons comparer le temps d'exécution d'un calcul simple via une boucle Python vs. NumPy.

### Exercice 5.1 : Comparaison du temps d'exécution

Nous allons calculer la somme des carrés de deux grands vecteurs de **10 millions** d'éléments. Utilisez la commande magique `%timeit` (dans les cellules Code) pour comparer les deux méthodes.

```python
# Créez deux très grands vecteurs (10 millions d'éléments)
N = 10_000_000
v_a = np.random.rand(N)
v_b = np.random.rand(N)

# --- Méthode 1: Utilisation d'une boucle 'for' Python --- 
print("Temps d'exécution avec la boucle 'for' (ATTENTION: sera lent!)")
# Décommentez les 3 lignes ci-dessous pour exécuter la mesure.
# %%timeit -n 1 -r 1 
# result = 0
# for i in range(N):
#     result += v_a[i]**2 + v_b[i]**2
```

```python
# --- Méthode 2: Utilisation de la vectorisation NumPy ---
print("Temps d'exécution avec la vectorisation NumPy:")
%%timeit 
np.sum(v_a**2 + v_b**2)
```

Le résultat de l'exercice 5.1 démontre de manière frappante que la **vectorisation NumPy est massivement plus rapide** (ordres de grandeur) que les boucles Python standard.
