## 3.2 Tableaux NumPy

La structure de données de base de NumPy est un *tableau* (array) indexable à n dimensions contenant des éléments du même type (`dtype`). Immédiatement, vous remarquerez peut-être que nous avons surchargé le terme 'dimension'. Ci-dessus, c'était le nombre d'éléments dans le vecteur, ici, la dimension fait référence au nombre d'index d'un tableau. Un tableau unidimensionnel ou 1-D a un seul index. Dans le Cours 1, nous représenterons les vecteurs comme des tableaux NumPy 1-D.

 - Tableau 1-D, forme (n,) : n éléments indexés de [0] à [n-1]

## 3.3 Création de Vecteurs


Les routines de création de données dans NumPy ont généralement un premier paramètre qui est la forme (shape) de l'objet. Cela peut être soit une valeur unique pour un résultat 1-D, soit un tuple (n,m,...) spécifiant la forme du résultat. Ci-dessous se trouvent des exemples de création de vecteurs utilisant ces routines.

```python
# Routines NumPy qui allouent de la mémoire et remplissent les tableaux avec des valeurs
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, forme de a = {a.shape}, type de données de a = {a.dtype}")
a = np.zeros((4,4));             print(f"np.zeros(4,4) :  a = {a}, forme de a = {a.shape}, type de données de a = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, forme de a = {a.shape}, type de données = {a.dtype}")
```

Certaines routines de création de données ne prennent pas de tuple de forme :

```python
# Routines NumPy qui allouent de la mémoire et remplissent les tableaux avec des valeurs mais n'acceptent pas la forme comme argument d'entrée
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, forme de a = {a.shape}, type de données = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, forme de a = {a.shape}, type de données = {a.dtype}")
```

Les valeurs peuvent aussi être spécifiées manuellement.

```python
# Routines NumPy qui allouent de la mémoire et remplissent avec des valeurs spécifiées par l'utilisateur
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     forme de a = {a.shape}, type de données = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, forme de a = {a.shape}, type de données = {a.dtype}")
```

Toutes ces commandes ont créé un vecteur unidimensionnel `a` avec quatre éléments. `a.shape` renvoie les dimensions. Ici, nous voyons a.shape = `(4,)` indiquant un tableau 1-d avec 4 éléments.

## 3.4 Opérations sur les Vecteurs
Explorons quelques opérations utilisant des vecteurs.
<a name="toc_40015_3.4.1"></a>
### 3.4.1 Indexation

**L'indexation** signifie faire référence à *un élément* d'un tableau par sa position dans le tableau.

NumPy commence l'indexation à zéro, donc le 3ème élément d'un vecteur $\mathbf{a}$ est `a[2]`.

```python
#opérations d'indexation vectorielle sur des vecteurs 1-D
a = np.arange(10)
print(a)

#accéder à un élément
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accéder à un élément renvoie un scalaire")

# accéder au dernier élément, les index négatifs comptent à partir de la fin
print(f"a[-1] = {a[-1]}")

#les index doivent être dans la plage du vecteur sinon ils produiront une erreur
try:
    c = a[10]
except Exception as e:
    print("Le message d'erreur que vous verrez est :")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 Découpage (Slicing)
**Le découpage (Slicing)** signifie obtenir un *sous-ensemble* d'éléments d'un tableau basé sur leurs indices.

Le slicing crée un tableau d'indices utilisant un ensemble de trois valeurs (`start:stop:step`). Un sous-ensemble de valeurs est également valide. Son utilisation est mieux expliquée par l'exemple :

```python
#opérations de découpage (slicing) vectoriel
a = np.arange(10)
print(f"a         = {a}")

#accéder à 5 éléments consécutifs (début:fin:pas)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# accéder à 3 éléments séparés par deux 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# accéder à tous les éléments à l'index 3 et au-dessus
c = a[3:];        print("a[3:]    = ", c)

# accéder à tous les éléments en dessous de l'index 3
c = a[:3];        print("a[:3]    = ", c)

# accéder à tous les éléments
c = a[:];         print("a[:]     = ", c)

# les 3 valeurs à partir du dernier élément
c = a[-3:];       print("a[-3:]   = ", c)
```

<a name="toc_40015_3.4.3"></a>
### 3.4.3 Opérations sur vecteur unique
Il existe un certain nombre d'opérations utiles qui impliquent des opérations sur un seul vecteur.

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# inverser le signe des éléments de a
b = -a 
print(f"b = -a        : {b}")

# somme de tous les éléments de a, renvoie un scalaire
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

<a name="toc_40015_3.4.4"></a>
### 3.4.4 Opérations élément par élément Vecteur-Vecteur
La plupart des opérations arithmétiques, logiques et de comparaison de NumPy s'appliquent également aux vecteurs. Ces opérateurs fonctionnent élément par élément. Par exemple
$$ c_i = a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Les opérateurs binaires fonctionnent élément par élément : {a + b}")
```

### 3.4.6 Produit scalaire Vecteur-Vecteur (Dot Product)
Le produit scalaire est un pilier de l'algèbre linéaire et de NumPy. C'est une opération utilisée de manière extensive dans ce cours et elle doit être bien comprise. Le produit scalaire est montré ci-dessous.

<img src="./images/C1_W2_Lab04_dot_notrans.gif" width=800> 

Le produit scalaire multiplie les valeurs de deux vecteurs élément par élément puis somme le résultat.
Le produit scalaire vectoriel nécessite que les dimensions des deux vecteurs soient les mêmes.

Implémentons notre propre version du produit scalaire ci-dessous :

**En utilisant une boucle for**, implémentez une fonction qui renvoie le produit scalaire de deux vecteurs. La fonction doit retourner pour les entrées données $a$ et $b$ :
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Supposez que `a` et `b` ont la même forme.

```python
def my_dot(a, b): 
    """
   Calcule le produit scalaire de deux vecteurs
 
    Args:
      a (ndarray (n,)):  vecteur d'entrée 
      b (ndarray (n,)):  vecteur d'entrée avec la même dimension que a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Notez que le produit scalaire est censé renvoyer une valeur scalaire.

Essayons les mêmes opérations en utilisant `np.dot`.

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

Ci-dessus, vous remarquerez que les résultats pour le 1-D correspondent à notre implémentation.

<a name="toc_40015_3.4.7"></a>
### 3.4.7 Le besoin de vitesse : vecteur vs boucle for
Nous avons utilisé la bibliothèque NumPy car elle améliore la vitesse et l'efficacité mémoire. Démontrons cela :

```python
np.random.seed(1)  #pour la reproductibilité des valeur aléatoires (génere les mêmes valeurs à chaque exécution)
a = np.random.rand(10000000)  # très grands tableaux
b = np.random.rand(10000000)

tic = time.time()  # capturer l'heure de début
c = np.dot(a, b)
toc = time.time()  # capturer l'heure de fin

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Durée version vectorisée: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capturer l'heure de début
c = my_dot(a,b)
toc = time.time()  # capturer l'heure de fin

print(f"my_dot(a, b) =  {c:.4f}")
print(f"Durée version boucle: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #supprimer ces grands tableaux de la mémoire
```

Ainsi, la vectorisation offre une grande accélération dans cet exemple. C'est parce que NumPy fait un meilleur usage du parallélisme des données disponible dans le matériel sous-jacent. Les GPU et les CPU modernes implémentent des pipelines SIMD (Single Instruction, Multiple Data) permettant à plusieurs opérations d'être émises en parallèle. Ceci est critique en Machine Learning où les ensembles de données sont souvent très volumineux.

## 4.4 Opérations sur les Matrices
Explorons quelques opérations utilisant des matrices.

<a name="toc_40015_4.4.1"></a>
### 4.4.1 Indexation


Les matrices incluent un second index. Les deux index décrivent [ligne, colonne]. L'accès peut renvoyer soit un élément, soit une ligne/colonne. Voir ci-dessous :

```python
#opérations d'indexation vectorielle sur des matrices
a = np.arange(6).reshape(-1, 2) #reshape est un moyen pratique de créer des matrices
print(f"a.shape: {a.shape}, \na= {a}")

#accéder à un élément
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accéder à un élément renvoie un scalaire\n")

#accéder à une ligne
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

Il vaut la peine d'attirer l'attention sur le dernier exemple. Accéder à une matrice en spécifiant juste la ligne renverra un *vecteur 1-D*.

**Reshape (Remodelage)** 
L'exemple précédent utilisait [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) pour façonner le tableau.  
`a = np.arange(6).reshape(-1, 2) `   
Cette ligne de code a d'abord créé un *vecteur 1-D* de six éléments. Elle a ensuite remodelé ce vecteur en un tableau *2-D* en utilisant la commande reshape. Cela aurait pu être écrit :  
`a = np.arange(6).reshape(3, 2) `  
Pour arriver au même tableau de 3 lignes et 2 colonnes.
L'argument -1 indique à la routine de calculer le nombre de lignes compte tenu de la taille du tableau et du nombre de colonnes.


<a name="toc_40015_4.4.2"></a>
### 4.4.2 Découpage (Slicing)
Le slicing crée un tableau d'indices utilisant un ensemble de trois valeurs (`start:stop:step`). Un sous-ensemble de valeurs est également valide. Son utilisation est mieux expliquée par l'exemple :

```python
#Rdimentionnement de tableaux 1-D et 2-D
a = np.arange(20)
print(f"a = \n{a}")

b = a.reshape(2, 10)
print(f"b = \n{b}")

c = b.reshape(-1, 10)  #-1 indique à NumPy de calculer automatiquement la taille de cette dimension
print(f"c = \n{c}")
```

```python
#opérations de découpage (slicing) vectoriel 2-D
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#accéder à 5 éléments consécutifs (début:fin:pas)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "un tableau 1-D")
```

```python
#opérations de découpage (slicing) vectoriel 2-D
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#accéder à 5 éléments consécutifs (début:fin:pas)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "un tableau 1-D")

#accéder à 5 éléments consécutifs... sur deux lignes
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "un tableau 2-D")

# accéder à tous les éléments
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# accéder à tous les éléments d'une ligne (usage très courant)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "un tableau 1-D")
# identique à
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "un tableau 1-D")
```
